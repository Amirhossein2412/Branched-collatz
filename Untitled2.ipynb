```python
import time
from numba import int64, njit
import numpy as np


@njit
def run_standard_collatz_1b(limit=1_000_000_000):
    # Explicit memoization table for total stopping time to 1
    total_stopping_time = np.zeros(limit + 1, dtype=np.uint32)

    max_steps = np.uint32(0)
    max_steps_seed = int64(1)

    min_steps = np.uint32(1000000)
    min_steps_seed = int64(2)

    max_peak_val = int64(1)
    max_peak_seed = int64(1)

    total_steps_sum = np.float64(0.0)
    MAX_STEPS_THRESHOLD = 50000

    for i in range(2, limit + 1):
        curr = int64(i)
        steps_to_drop = np.uint32(0)
        local_peak = curr

        while curr >= i:
            if curr % 2 == 0:
                curr = curr // 2
            else:
                curr = 3 * curr + 1

            if curr > local_peak:
                local_peak = curr

            steps_to_drop += 1
            if steps_to_drop > MAX_STEPS_THRESHOLD:
                return (
                    False,
                    max_steps,
                    max_steps_seed,
                    min_steps,
                    min_steps_seed,
                    max_peak_val,
                    max_peak_seed,
                    0.0,
                )

        # Dynamic programming relation: T(i) = steps_to_drop + T(curr)
        exact_total_steps = steps_to_drop + total_stopping_time[curr]
        total_stopping_time[i] = exact_total_steps
        total_steps_sum += exact_total_steps

        if exact_total_steps > max_steps:
            max_steps = exact_total_steps
            max_steps_seed = i

        if exact_total_steps < min_steps:
            min_steps = exact_total_steps
            min_steps_seed = i

        if local_peak > max_peak_val:
            max_peak_val = local_peak
            max_peak_seed = i

        if i % 100_000_000 == 0:
            print("Milestone reached:", i)

    avg_steps = total_steps_sum / (limit - 1)
    return (
        True,
        max_steps,
        max_steps_seed,
        min_steps,
        min_steps_seed,
        max_peak_val,
        max_peak_seed,
        avg_steps,
    )


def test_standard_collatz():
    limit = 1_000_000_000
    print("Testing Standard Collatz (1 Billion)...")
    start_time = time.time()

    (
        success,
        max_steps,
        max_steps_seed,
        min_steps,
        min_steps_seed,
        max_peak_val,
        max_peak_seed,
        avg_steps,
    ) = run_standard_collatz_1b(limit)

    total_time = time.time() - start_time

    if not success:
        print("Verification failed: divergence or cycle detected.")
        return

    print("\n" + "=" * 55)
    print("STANDARD COLLATZ SUMMARY (1 BILLION)")
    print("=" * 55)
    print(f"Tested range                  : 1 to {limit:,}")
    print(f"Total execution time          : {total_time:.2f} seconds")
    print(f"Minimum steps to 1 (for n > 1): {min_steps} (seed: {min_steps_seed:,})")
    print(f"Maximum steps to 1            : {max_steps} (seed: {max_steps_seed:,})")
    print(f"Average total steps to 1      : {avg_steps:.4f}")
    print(
        f"Highest peak value reached    : {max_peak_val:,} (seed: {max_peak_seed:,})"
    )
    print("=" * 55)


if __name__ == "__main__":
    test_standard_collatz()
```
